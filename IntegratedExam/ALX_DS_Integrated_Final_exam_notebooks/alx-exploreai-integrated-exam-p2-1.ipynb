```python
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error
from sklearn.metrics import r2_score
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.linear_model import Ridge, Lasso
```

```python
# Load the dataset
df = pd.read_csv('/kaggle/input/alx-final-exam-rand-roller/rand-dollar.csv', encoding='ISO-8859-1', index_col=0)
df
```

**What is the value of the intercept of the model?**

* 8.67 e-5
* -1.24
* 3.99
* 3.29

```python
# Separate features and target
X = df[['Value of Exports (ZAR)']]  # Predictor variable
y = df['ZAR/USD']  # Target variable

# Split the data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, shuffle=False)

# Train the linear regression model
lm = LinearRegression()
lm.fit(X_train, y_train)

# Get the intercept of the model
intercept = lm.intercept_

print(f"The value of the intercept is: {intercept}")
```

```python
# Retrieve the coefficient (slope) of the model
slope = lm.coef_[0]

print(f"The value of the slope is: {slope}")
```

**What is the MSE of the model on the test set?**

* -8.45
* 6
* 1.89
* 8.22

```python
y_pred = lm.predict(X_test[['Value of Exports (ZAR)']])
y_pred
```

```python
mse = mean_squared_error(y_test, y_pred)
mse
```

**What is the R-squared value of the model on the test set?**

* 9

**Which variable has the weakest linear relationship with the ZAR/USD exchange rate?**

* Consumer Price Index
* Claims on Non-residents
* Savings Rate
* Lending Rate

```python
# Identify non-numeric columns
non_numeric_columns = df.select_dtypes(exclude=[float, int]).columns
print("Non-numeric columns:", non_numeric_columns)

# One-hot encode categorical columns
df_encoded = pd.get_dummies(df, columns=non_numeric_columns, drop_first=True)

# Calculate correlations
correlations = df_encoded.corr()
```

```python
# Extract correlations with the target variable 'ZAR/USD'
target_correlation = correlations['ZAR/USD']

# Drop 'ZAR/USD' from the series to avoid comparing it with itself
target_correlation = target_correlation.drop('ZAR/USD')

# Find the variable with the weakest (lowest) correlation
weakest_correlation = target_correlation.idxmin()
weakest_correlation_value = target_correlation[weakest_correlation]

print(f"The variable with the weakest linear relationship with ZAR/USD is {weakest_correlation} with a correlation of {weakest_correlation_value}.")
```

**Which variable has the strongest linear relationship with the ZAR/USD exchange rate?**

* Consumer Price Index
* Claims on Non-residents
* Savings Rate
* IMF Reserve Position (USD)

```python
# Find the variable with the strongest (highest) correlation
strongest_correlation = target_correlation.idxmax()
strongest_correlation_value = target_correlation[strongest_correlation]

print(f"The variable with the strongest linear relationship with ZAR/USD is {strongest_correlation} with a correlation of {strongest_correlation_value}.")
```

Split the original DataFrame into X (features) and y (targets)
Standardise the entire X matrix
Create X_train, X_test, y_train, y_test using the same chronological 80/20 split as before
Train two models, “ridge” and “lasso”, which use ridge regression and LASSO, respectively (in the case of the lasso model, set alpha=0.01, use default parameters for the ridge model)
**What is the training MSE of the Ridge model?**

* 0.632
* 0.040
* 0.579
* 0.047

```python
# Split into features (X) and target (y)
X = df.drop(columns=['ZAR/USD'])
y = df['ZAR/USD']

# Standardize the features
scaler = StandardScaler()
X_standardized = scaler.fit_transform(X)

# Create chronological 80/20 split
X_train, X_test, y_train, y_test = train_test_split(X_standardized, y, test_size=0.2, shuffle=False)
```

```python
# Train Ridge Regression model
ridge_model = Ridge()
ridge_model.fit(X_train, y_train)

# Train Lasso Regression model
lasso_model = Lasso(alpha=0.01)
lasso_model.fit(X_train, y_train)
```

```python
# Make predictions on the training set
y_train_pred_ridge = ridge_model.predict(X_train)

# Calculate the training MSE for Ridge model
training_mse_ridge = mean_squared_error(y_train, y_train_pred_ridge)
print(f"Training MSE of the Ridge model: {training_mse_ridge:.3f}")
```

```python
# Make predictions on the training set for LASSO model
y_train_pred_lasso = lasso_model.predict(X_train)

# Calculate the training MSE for LASSO model
training_mse_lasso = mean_squared_error(y_train, y_train_pred_lasso)
print(f"Training MSE of the LASSO model: {training_mse_lasso:.3f}")
```

```python
# Make predictions on the test set for Ridge model
y_test_pred_ridge = ridge_model.predict(X_test)

# Calculate the testing MSE for Ridge model
testing_mse_ridge = mean_squared_error(y_test, y_test_pred_ridge)
print(f"Testing MSE of the Ridge model: {testing_mse_ridge:.3f}")
```

```python
# Train LASSO model with alpha=0.01
lasso_model = Lasso(alpha=0.01)
lasso_model.fit(X_train, y_train)

# Make predictions on the test set for LASSO model
y_test_pred_lasso = lasso_model.predict(X_test)

# Calculate the testing MSE for LASSO model
testing_mse_lasso = mean_squared_error(y_test, y_test_pred_lasso)
print(f"Testing MSE of the LASSO model: {testing_mse_lasso:.3f}")
```

```python
# Create a DataFrame to view coefficients alongside feature names
coefficients_df = pd.DataFrame({
    'Feature': X.columns,
    'Coefficient': ridge_coefficients
})

# Sort by absolute value of coefficients
coefficients_df['Abs_Coefficient'] = coefficients_df['Coefficient'].abs()
coefficients_df = coefficients_df.sort_values(by='Abs_Coefficient', ascending=False)

print(coefficients_df)
```
